```python
# Before working with the data,
# we download and prepare all features

# load all examples from the file
data = np.genfromtxt('winequality-white.csv',delimiter=";",skip_header=1)

print(f"{data.shape = }")

# Prepare for proper training
rng = np.random.default_rng(1234)
rng.shuffle(data)  # randomly sort examples

# take the first 3000 examples for training
# (remember array slicing from last week)
X_train = data[:3000, :11]  # all features except last column
y_train = data[:3000, 11]  # quality column

# and the remaining examples for testing
X_test = data[3000:, :11]  # all features except last column
y_test = data[3000:, 11]  # quality column

print("First example:")
print("Features:", X_train[0])
print("Quality:", y_train[0])
```

# Hints

Formally, we want to find weights $w_i$ that minimize:
$$
\sum_{j}\left(\sum_{i} X_{i j} w_{i}-y_{j}\right)^{2}
$$
The index $i$ denotes the different features (properties of the wines) while the index $j$ runs over the different wines. The matrix $X_{ij}$ contains the training data, $y_j$ is the 'true' quality for sample $j$. The weights can be found by taking the first derivative of the above expression with respect to the weights and setting it to zero (the standard strategy for finding an extremum), and solving the corresponding system of equations (for a detailed derivation, see [here](https://en.wikipedia.org/wiki/Ordinary_least_squares)). The result is:
$$
\overrightarrow{\mathbf{w}}=\left(\mathbf{X}^{T} \mathbf{X}\right)^{-1} \mathbf{X}^{T} \overrightarrow{\mathbf{y}}
$$

In the end, you should have as many components of $w_i$ as there are features in the data (i.e. eleven in this case).

You can use `.shape` to inspect the dimensions of numpy tensors.


```python
feature_names = [
    "fixed acidity",
    "volatile acidity",
    "citric acid",
    "residual sugar",
    "chlorides",
    "free sulfur dioxide",
    "total sulfur dioxide",
    "density",
    "pH",
    "sulphates",
    "alcohol"
]

fig, axes = plt.subplots(3, 4, figsize=(16, 10))
for i, ax in enumerate(axes.flat[:11]):
    ax.hist(X_train[:, i], bins=30)
    ax.set_title(feature_names[i])
axes.flat[11].hist(y_train, bins=np.arange(2.5, 10.5, 1))   # bonus: the target itself
axes.flat[11].set_title("quality")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(3, 4, figsize=(16, 10))
for i, ax in enumerate(axes.flat[:11]):
    ax.scatter(X_train[:, i], y_train, s=5, alpha=0.2)
    ax.set_xlabel(feature_names[i])
    ax.set_ylabel("quality")
axes.flat[11].axis("off")
plt.tight_layout()
plt.show()
```

**Comment on the data.** Quality is an integer score and most wines are rated 5–7, very few 3–4 or 8–9. The features live on very different scales (total sulfur dioxide goes up to about 300, density is about 0.99–1.04), and some of them (residual sugar, chlorides, free sulfur dioxide) are skewed with a few outliers. Because quality only takes integer values, the scatter plots show horizontal stripes, so I used small transparent points to see where most wines are.

```python
corrs = np.array([np.corrcoef(X_train[:, i], y_train)[0, 1] for i in range(11)])
for name, r in zip(feature_names, corrs):
    print(f"{name:22s} {r:+.3f}")
print("Most predictive:", feature_names[np.argmax(np.abs(corrs))])
```

**Which feature is most predictive?** Alcohol has the strongest correlation with quality (r ≈ +0.43), so by itself it is the most predictive feature. Density is second (r ≈ −0.30), but density and alcohol are strongly related (alcohol is lighter than water), so density partly carries the same information. All other features are only weakly correlated with quality on their own. I ranked by |r|, since a strong negative correlation is just as informative as a positive one.

```python
w = np.linalg.inv(X_train.T @ X_train) @ X_train.T @ y_train
print("w.shape =", w.shape)
for name, wi in zip(feature_names, w):
    print(f"{name:22s} {wi:+.4f}")
```

**Comment on the weights.** The formula $\vec w = (X^TX)^{-1}X^T\vec y$ gives 11 weights, one per feature, as expected. The weights cannot be compared directly with each other, because the features have different units and scales. The model has no bias term, so it is forced through the origin, adding a column of ones to $X$ would give a 12th weight acting as an intercept.

```python
y_pred = X_test @ w
mse = np.mean((y_pred - y_test)**2)
r_test = np.corrcoef(y_pred, y_test)[0, 1]
print(f"Test MSE: {mse:.3f}")
print(f"Baseline MSE (always predict training mean): {np.mean((y_train.mean() - y_test)**2):.3f}")
print(f"Correlation predicted vs true: {r_test:.3f}")
```

```python
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred, s=8, alpha=0.3)
plt.plot([3, 9], [3, 9], "k--", label="perfect prediction")
plt.xlabel("true quality")
plt.ylabel("predicted quality")
plt.title("Test set: predicted vs true quality")
plt.legend()
plt.show()
```

**How does the prediction compare with the true quality?** The correlation between predicted and true quality on the test set is about 0.52 and the test MSE is about 0.57, compared with about 0.77 for always predicting the mean. So the model helps, but $r^2 \approx 0.27$ means it only explains about a quarter of the variation. In the scatter plot the predictions are squeezed toward the middle (about 5–7): bad wines (3–4) are predicted too high and very good wines (8–9) too low. Likely reasons are that quality is a subjective, coarse integer score, the true relation is probably not linear, and things like taste and aroma are not in the chemical measurements.
